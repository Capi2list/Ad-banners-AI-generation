```python
import pandas as pd


train_data = pd.read_csv('TrainData.csv')
test_data = pd.read_csv('TestData.csv')

print(train_data.isna().sum())
```

```python
train_data['0'].mean()
```

```python
train_data = train_data.fillna({'1': train_data['1'].mean()})
```

```python
train_data = train_data.drop([347, 364, 381])
```

```python
train_data['target'] = train_data['target'].replace('Chorn', 'Churn')
train_data['target'] = train_data['target'].replace('Not chorn', 'Not churn')
```

```python
train_data['target'] = (train_data['target'] == 'Churn')
```

```python
train_data[(train_data['2'] > train_data['2'].mean()) & (train_data['13'] < train_data['13'].median())]
```

```python
from sklearn.preprocessing import LabelEncoder

LE = LabelEncoder()
train_data['7'] = LE.fit_transform(train_data['7'])
```

```python
test_data['7'] = LE.transform(test_data['7'])
```

```python
test_data['7'].max()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score
import numpy as np

X_train, y_train = train_data.drop('target', axis=1), train_data['target']
logreg = LogisticRegression(random_state = 42)

print(np.mean(cross_val_score(logreg, X_train, y_train, scoring='f1', cv=3)))
```

```python
from sklearn.model_selection import GridSearchCV

logreg = LogisticRegression(random_state = 42)
clf = GridSearchCV(logreg, param_grid = {'C' : [0.001,0.01,0.1,1,10,100]}, scoring='f1', cv=3)
clf.fit(X_train, y_train)
```
